# 行为时间轴 + 三视角视频合成示例

本 notebook 演示如何使用 `scripts.analysis.render_behavior_timeline_video` 中的函数，
将三视角原始视频（top / side1 / side2）与 BORIS 导出的行为事件 CSV 叠加，
生成一个两行布局的新视频：

- 第一行：top / side1 / side2 三路视频并排显示；
- 第二行：整段实验时长的行为时间轴：
  - 颜色编码不同行为（及执行该行为的个体），
  - 下方图例说明颜色对应的“个体: 行为”，
  - 一条随时间移动的光标，时间格式为 mm:ss:ff。

```python
from pathlib import Path

from scripts.analysis.render_behavior_timeline_video import render_three_view_with_behavior_timeline

# 根据自己数据修改这里的路径
video_top = Path(r"D:\videos_raw\20251225_K_1_2_FCMC\top_cam\cam0_top.mp4")
boris_csv = Path(r"D:\boris_events\20251225_K_1_2_FCMC.csv")
output = Path(r"D:\videos_output\20251225_K_1_2_FCMC_behavior_timeline.mp4")
```

```python
# 只导出指定帧区间的片段（推荐先用小范围预览效果）
# 例如：start_frame = 0, end_frame = 19 → 前 20 帧
start_frame = 60000
end_frame = 79999

segment_output = output.with_name(output.stem + f"_f{start_frame}_{end_frame}.mp4")

render_three_view_with_behavior_timeline(
    video_top=video_top,
    boris_csv=boris_csv,
    output_path=segment_output,
    start_frame=start_frame,
    end_frame=end_frame,
)
segment_output
```

```python
render_three_view_with_behavior_timeline(
    video_top=video_top,
    boris_csv=boris_csv,
    output_path=output,
)
output
```
